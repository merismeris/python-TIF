# Funciones

Las funciones se usan para organizar el flujo del programa, especialmente para permitirnos realizar tareas comunes repetidamente de forma sencilla. Ya hemos usado muchas funciones, como las usadas a través del módulo `math`, integradas en métodos de clases como las que operan sobre listas (`append()` y `pop()`) o sobre cadenas (como `split()`). Aquí veremos cómo escribir nuestras propias funciones.

Para poder usar nuestras propias funciones, primero tenemos que definirlas. Podemos definirla en el propio programa o en módulos externos que luego importamos. Veremos ambas opciones en prácticas. 

## Definición de funciones

Para definir una función debemos:
- Usar la palabra clave `def`, seguida del nombre de la función y paréntesis `()`.
- Dentro de los paréntesis, podemos listar los parámetros que la función tomará como entrada.
- Terminar la línea de definición con dos puntos `:`.
- Identar el bloque de código que constituye el cuerpo de la función.
- Incluir una declaración de retorno para devolver un valor usando la palabra clave `return`. Si no se incluye una declaración de retorno, la función devolverá `None` por defecto.

Aquí hay un ejemplo simple de una función que toma un único argumento, `i`

````python
def my_fun(i):
    print(f"Recibido el valor i = {i}")
````

Para definir una función en un módulo externo, simplemente escribimos la definición de la función en un archivo `.py` separado. Por ejemplo, podemos crear un archivo llamado `mimodulo.py` con la siguiente función:

```python
def square(i):
    return i * i
```

### Llamada a funciones


Para llamar a la función, simplemente usamos su nombre seguido de paréntesis que contienen los argumentos que queremos pasarle. Por ejemplo:

````python
my_fun(10)
my_fun(5)
````

```{note}
Podemos definir una función en cualquier parte del código (siempre **antes de llamarla**), pero es una buena práctica definirlas al principio del archivo o en módulos separados para mantener el código organizado. 
```

Si hemos creado una función en un módulo externo, podemos importarla en nuestro programa principal usando la instrucción `import` o `from ... import ...`. 

En el ejemplo anterior, podemos importar la función `square` y usarla en nuestro programa principal de la siguiente manera:
```python
from mimodulo import square

print(square(5))  # Salida: 25
```

En cualquier caso, las funciones siempre devuelven un valor (o varios) de cualquier tipo. Si no se da explícitamente, devuelven `None` (el equivalente a `null` en otros lenguajes).

````python
a = my_fun(10)
a
````

Aquí hay una función simple que toma dos números y devuelve su producto.

````python
def multiply(a, b):
    return a*b

c = multiply(3, 4)
c
````

```{note}
Las funciones son un lugar donde entra en juego el _alcance_ (scope). Una función tiene su propio _namespace_. Si una variable no está definida en esa función, buscará en el namespace desde donde fue llamada para ver si esa variable existe allí.

Sin embargo, deberías evitar esto tanto como sea posible, e intentar que todas las variables que necesite la función sean variables locales (las variables que persisten a través de namespaces se llaman variables globales).
```

## Mutabiliad e inmutabilidad de los argumentos

Cuando pasamos argumentos a una función, estos pueden ser mutables o inmutables. En función de esta característica, Python maneja los argumentos de manera diferente.

- En el caso de los argumentos inmutables (como enteros, cadenas y tuplas), Python crea una copia del valor y lo pasa a la función. Cualquier cambio realizado en el argumento dentro de la función no afectará al valor original fuera de la función. Esto se denomina **paso por valor**.
- En el caso de los argumentos mutables (como listas y diccionarios), Python pasa una referencia al objeto original. Esto significa que cualquier cambio realizado en el argumento dentro de la función afectará al objeto original fuera de la función. Esto se denomina **paso por referencia**. Os dejamos un ejemplo para ilustrar esto:



````python
def agregar_elemento(lista, elemento):
    lista.append(elemento)
    return lista
mi_lista = [1, 2, 3]
nueva_lista = agregar_elemento(mi_lista, 4)
print(mi_lista)      # Salida: [1, 2, 3, 4]
print(nueva_lista)   # Salida: [1, 2, 3, 4] 
````

```{important} 
El paso por referencia o valor puede llevar a errores difíciles de detectar si no se sabe bien qué objetos son mutables y cuáles no. Para saber más sobre mutabilidad e inmutabilidad en Python, podéis consultar la [documentación oficial](https://docs.python.org/3/reference/datamodel.html#objects-values-and-types).
```

## Argumentos por palabra clave

Puedes tener argumentos opcionales que proveen valores por defecto. Aquí hay una función simple que calcula $\sin(\theta)$ donde $\theta$ opcionalmente puede estar en grados.

````python
import math
````

````python
def my_sin(theta, in_degrees=False):
    if in_degrees:
        return math.sin(math.radians(theta))
    return math.sin(theta)
````

````python
my_sin(math.pi/4), my_sin(45, in_degrees=True)                    
````

```{important}
Es importante destacar que Python evalúa los argumentos opcionales una sola vez—cuando la función se define. Esto significa que si pones como valor por defecto un objeto mutable vacío, por ejemplo, persistirá entre llamadas. Esto tiene que ver con la mutabilidad que hemos comentado antes. 

**Esto conduce a uno de los errores más comunes entre quienes empiezan**
```

Aquí hay un ejemplo de intentar inicializar con una lista vacía:

````python
def f(a, L=[]):
    L.append(a)
    return L
````

````python
print(f(1))
print(f(2))
print(f(3))
````

Fíjate que cada llamada no crea su propia lista separada. En su lugar, se creó una única lista vacía cuando la función fue procesada por primera vez, y esta lista persiste en memoria como el valor por defecto para el argumento opcional `L`.

Si queremos que se cree una lista única cada vez (por ejemplo, un lugar separado en memoria), en su lugar inicializamos el argumento a `None` y luego comprobamos su valor real y creamos una lista vacía en el cuerpo de la función si el valor por defecto no cambió.

````python
def fnew(a, L=None):
    if L is None:
        L = []
    L.append(a)
    return L
````

````python
print(fnew(1))
print(fnew(2))
print(fnew(3))
````

````python
L = fnew(1)
print(fnew(2, L=L))
````

Fíjate que el mismo `None` que vimos anteriormente vuelve a aparecer aquí.

````python
L
````

## Lambdas (funciones anónimas)

Las lambdas son funciones "desechables". Son funciones pequeñas y sin nombre que a menudo se usan como argumentos en otras funciones.

Por ejemplo,  tenemos una lista de tuplas. Queremos ordenar la lista según el segundo elemento de la tupla. El método `sort` puede tomar un argumento opcional `key` que nos indica cómo interpretar el elemento de la lista para ordenar.

````python
pairs = [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]
pairs.sort(key=lambda p: p[1])
````

````python
pairs
````

Aquí usamos una lambda en una extracción de una lista (con la función `filter`)

````python
squares = [x**2 for x in range(100)]
sq = list(filter(lambda x : x%2 == 0 and x%3 == 0, squares))
sq
````
