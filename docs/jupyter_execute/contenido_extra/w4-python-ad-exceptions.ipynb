# Excepciones

Python lanza excepciones cuando encuentra un error. La idea es que se pueden capturar estas excepciones y tomar una acción apropiada en lugar de que el código falle. El mecanismo para esto es `try` / `except`. A continuación se muestra un ejemplo que causa una excepción `ZeroDivisionError` (Error de División por Cero):

```python
a = 1/0
```

A continuación se muestra cómo manejar esta excepción:

```python
try:
    a = 1/0
except ZeroDivisionError:
    print("advertencia: has dividido por cero")
    a = 1
a
```

Otro ejemplo: intentar acceder a una clave que no existe en un diccionario:

```python
dict = {"a":1, "b":2, "c":3}
print(dict["d"])
```

`KeyError` es la excepción que se lanzó. Podemos capturar esta excepción y tomar la acción apropiada:

```python
try:
    val = dict["d"]
except KeyError:
    val = None

print(val)
```

Todavía no hemos hablado de E/S (entrada/salida), pero cuando se abre un archivo, se obtiene un objeto con el que se puede interactuar. A continuación se muestra un ejemplo de intento de abrir un archivo para lectura que no existe:

```python
try:
    f = open("file.txt", "r")
except IOError:
    print("error: ese archivo no existe")
```

Existen muchos tipos diferentes de excepciones que se pueden capturar, y es posible capturar múltiples excepciones por cláusula `except` o tener múltiples cláusulas `except`. Probablemente no se podrán anticipar todos los modos de fallo por adelantado. En ese caso, cuando se ejecute el código y falle debido a una excepción, el intérprete de Python imprimirá el nombre de la excepción, lo que permitirá modificar el código para tomar la acción apropiada.
