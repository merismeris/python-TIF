## Comentarios

En Python, los comentarios se indican con el símbolo `#`. Todo lo que siga a este símbolo en una línea será ignorado por el intérprete de Python. Los comentarios son útiles para explicar el código y hacer que sea más legible.

```python
# Esto es un comentario en Python
print("Hola, mundo!")  # Esto también es un comentario
```
También se permiten "comentarios" multilínea utilizando comillas triples (`'''` o `"""`). Estos no son técnicamente comentarios, pero se usan comúnmente para documentar bloques de código o funciones:

```python
'''
Esto es un comentario
multilínea en Python
'''

Al igual que sucede en otros lenguajes, existen convenciones para escribir comentarios claros y útiles. Además existen notaciones para documentar funciones y módulos, como las docstrings (en Python, cadenas de documentación). Ejemplo documentando una función:

```python
def suma(a, b):
    """
    Esta función devuelve la suma de dos números.

    Args:
        a (int | float): El primer número a sumar.
        b (int | float): El segundo número a sumar.

    Returns:
        int | float: La suma de a y b.
    """
    return a + b
```

## Identificadores y variables
Los identificadores son los nombres que se utilizan para identificar variables, funciones, clases y otros objetos en Python. Hay algunas reglas y convenciones que se deben seguir al nombrar identificadores:
- Deben comenzar con una letra (a-z, A-Z) o un guion bajo (_), seguido de letras, dígitos (0-9) o guiones bajos.
    - Los guiones bajos al principio suelen tener un significado especial (por ejemplo, `_variable` indica que es una variable "privada").
- No pueden ser iguales a las palabras reservadas del lenguaje.
- Son sensibles a mayúsculas y minúsculas (por ejemplo, `variable` y `Variable` son diferentes).
- Se recomienda utilizar nombres descriptivos y significativos para mejorar la legibilidad del código.
- Se recomienda utilizar el estilo de nomenclatura `snake_case` para variables y funciones (por ejemplo, `mi_variable`, `mi_funcion`), y `PascalCase` (o `CapWords`) para las clases (por ejemplo, `MiClase`).

```python
# Notación recomendada
mi_variable = 10

# Notación recomendada cuando es una Constante
PI = 3.1416

# Notación recomendada para funciones
def mi_funcion():
    pass
``` 
Una variable es la representación de un dato cuyo valor puede variar durante la ejecución de un programa. En Python, no es necesario declarar el tipo de una variable antes de usarla, ya que Python es un lenguaje de tipado dinámico. Simplemente se asigna un valor a una variable utilizando el operador de asignación `=`, y Python infiere automáticamente el tipo de dato. Esto es, si asignamos un número entero a una variable, Python la tratará como un entero; si asignamos una cadena de texto, la tratará como una cadena, y así sucesivamente. Una variable puede cambiar de tipo durante la ejecución del programa si se le asigna un valor de un tipo diferente.
Es posible saber el tipo de dato de una variable utilizando la función `type()`:

````python
mi_variable = 10
print(type(mi_variable))  # Salida: <class 'int'>  

variable_flotante = 3.14
print(type(variable_flotante))  # Salida: <class 'float'>
variable_cadena = "Hola"
print(type(variable_cadena))  # Salida: <class 'str'>   
````

## Sentencia import
El núcleo de Python incluye un conjunto de funcionalidades básicas. Sin embargo, para tareas más especializadas, Python cuenta con un vasto ecosistema de **módulos**, **paquetes** y **bibliotecas** que extienden sus capacidades.

-   Un **módulo** es un archivo de Python (`.py`) que contiene definiciones y declaraciones (variables, funciones, clases). Permiten organizar el código de forma lógica y reutilizable.
-   Un **paquete** es una colección de módulos.
-   Una **biblioteca** es un término más general que se refiere a una colección de paquetes y módulos con un propósito común (por ejemplo, `NumPy` para computación numérica).

Python incluye una **biblioteca estándar** con muchos módulos útiles que están disponibles sin necesidad de instalación adicional (por ejemplo, `math`, `random`).

Para utilizar funcionalidades que no están en la biblioteca estándar (bibliotecas de terceros), generalmente se siguen estos pasos:

1.  **Instalación**: Se instalan en el entorno de Python usando el gestor de paquetes `pip`. Por ejemplo: `pip install nombre_paquete`.
2.  **Importación**: Se utiliza la sentencia `import` para hacer que el módulo o paquete esté disponible en nuestro código.
3.  **Uso**: Se utilizan las funciones, clases o variables proporcionadas por el módulo importado.

````python
import math  # Importa el módulo math
resultado = math.sqrt(16)  # Usa la función sqrt del módulo math
print(resultado)  # Salida: 4.0
````

Existen varias formas de utilizar la sentencia `import`:

| Sintaxis                                  | Descripción                                                                                                                            |
|:------------------------------------------|:---------------------------------------------------------------------------------------------------------------------------------------|
| `import modulo`                           | Importa el módulo completo. Se accede a sus miembros con `modulo.miembro`.                                                             |
| `import modulo as alias`                  | Importa el módulo y le asigna un alias. Útil para acortar nombres largos (`import numpy as np`). Se accede con `alias.miembro`.          |
| `from modulo import miembro`              | Importa un miembro específico (función, clase, etc.). Se puede usar directamente (`miembro`) sin el prefijo del módulo.                  |
| `from modulo import miembro as alias`     | Importa un miembro específico y le asigna un alias.                                                                                    |
| `from modulo import miembro1, miembro2`   | Importa múltiples miembros específicos del mismo módulo.                                                                               |
| `from modulo import *`                    | Importa todos los miembros públicos del módulo. **No se recomienda** porque puede crear conflictos de nombres y reducir la legibilidad. |


````python

from math import sqrt  # Importa solo la función sqrt del módulo math
resultado = sqrt(25)  # Usa la función sqrt directamente
print(resultado)  # Salida: 5.0
````

Podemos importar varios módulos en una sola línea separándolos con comas, aunque la guía de estilo de Python (PEP 8) recomienda una importación por línea:

````python
# Forma válida pero no recomendada
import math, random

# Forma recomendada (PEP 8)
import math
import random

numero_aleatorio = random.randint(1, 10)  # Usa la función randint del módulo random
print(numero_aleatorio)  # Salida: un número aleatorio entre 1 y 10

resultado = math.factorial(5)  # Usa la función factorial del módulo math
print(resultado)  # Salida: 120
````

Para conocer las funciones que contiene un módulo, se puede utilizar la función `dir()`. Y para conocer cómo se usa una función específica, se puede utilizar la función `help()`.

````python
import math
print(dir(math))  # Muestra todas las funciones y variables del módulo math
help(math.sqrt)  # Muestra la documentación de la función sqrt del módulo math
````
