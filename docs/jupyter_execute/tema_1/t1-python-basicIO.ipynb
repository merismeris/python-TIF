# Entrada y salida de datos (por consola)
Uno de los primeros pasos que se realizan al comenzar con un lenguaje de programación es aprender a manejar la entrada y salida de datos. La primera forma de entrada y salida de datos que veremos será a través de la consola o terminal, es decir, la entrada y salida estándar.
Esto permitirá al usuario interactuar con el programa de forma sencilla, introduciendo datos y recibiendo resultados directamente en la consola. Más adelante, exploraremos otras formas de entrada y salida de datos, como archivos y bases de datos, pero por ahora nos centraremos en la consola para entender los conceptos básicos.
Muchos de estos conceptos son similares en otros lenguajes de programación (como MATLAB, C, Java, etc.), por lo que no debería ser difícil aprenderlos.

## input()
La función `input()` permite leer datos desde la entrada estándar (normalmente el teclado). Esta función toma un argumento opcional, que es una cadena que se muestra como un mensaje para el usuario. 

````python
x=input("Por favor, introduzca su nombre: ")
print("Hola, " + x + "!")
````

```{admonition} Advertencia
:class: warning
`input` siempre **devuelve una cadena**, por lo que si queremos leer otro tipo de dato, debemos convertirlo explícitamente. Esta operación se conoce comúnmente como *casting* o, de forma más precisa, **conversión de tipo**. Por ejemplo, para leer un número entero:
```

````python
x=input("Por favor, introduzca un número entero: ")
num=int(x)
print("El número que ha introducido es:", num)
````

O bien, más corto:

````python
num=int(input("Por favor, introduzca un número entero: "))
print("El número que ha introducido es:", num)
````

## print()
La función `print()` se utiliza para mostrar datos en la salida estándar (normalmente la pantalla). Puede tomar múltiples argumentos separados por comas, y los convertirá a cadenas si no lo son. Por ejemplo:


````python
print("El número que ha introducido es:", num)
````

Todos los argumentos que introduzcas en `print()` serán convertidos a cadenas y separados por espacios por defecto. Puedes cambiar el separador usando el parámetro `sep` y el final de línea con el parámetro `end`. Por ejemplo:

````python
print("Hola", "mundo", sep=", ", end="!\n") # Esto imprimirá `Hola, mundo!` en la pantalla.
````

Existen varias maneras de dar formato a cadenas en Python. Las más comunes son:

- **f-strings (cadenas literales formateadas)**: Disponibles en Python 3.6 y posteriores. Permiten incluir expresiones dentro de llaves `{}` que serán evaluadas en tiempo de ejecución. Se debe prefijar la cadena con una `f`.
- **Método `str.format()`**: Se usan llaves `{}` como marcadores de posición en la cadena y luego se llama al método `format()` con los valores a insertar.
- **Operador `%`**: Es una forma más antigua. Utiliza descriptores de formato como `%s` para cadenas y `%d` para enteros, y luego se proporciona una tupla con los valores.

Aquí tienes ejemplos de cada método:

````python
name = "Alice"
age = 30

print(f"Mi nombre es {name} y tengo {age} años.")#f-string. ¡Ojo a la f que se pone delante! 
print("Mi nombre es {name} y tengo {age} años.") # Esto no funcionará como se espera
print("Mi nombre es {} y tengo {} años.".format(name, age)) #str.format()
print("Mi nombre es %s y tengo %d años." % (name, age)) #operador % y descriptores de formato
````
