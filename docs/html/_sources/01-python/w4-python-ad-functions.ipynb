# Funciones lambda (funciones anónimas)

Las funciones lambda son pequeñas funciones anónimas que se definen utilizando la palabra clave `lambda`. Estas funciones pueden tener cualquier número de argumentos, pero solo una expresión. La sintaxis básica de una función lambda es la siguiente:

```python
lambda argumentos: expresión

```
Las funciones lambda se guardan en variables o se pasan directamente como argumentos a otras funciones. No requieren una declaración `def` y no tienen un nombre explícito.

Por ejemplo, podemos definir una función lambda que calcule el cuadrado de un número de la siguiente manera:


````python
square = lambda x: x ** 2
print(square(5))  # Salida: 25
````

`square` es una variable que contiene la función lambda. Es de tipo función y puede ser llamada como cualquier otra función en Python.
```python

````python
print(type(square))  # Salida: <class 'function'>
````

Las funciones lambda sirven para crear funciones pequeñas y desechables, generalmente utilizadas como argumentos para otras funciones, como `map()`, `filter()` y `sorted()`.

Por ejemplo,  tenemos una lista de tuplas. Queremos ordenar la lista según el segundo elemento de la tupla. El método `sort` puede tomar un argumento opcional `key` que nos indica cómo interpretar el elemento de la lista para ordenar.

````python
pairs = [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]
pairs.sort(key=lambda p: p[1])
pairs
````

El código anterior con sort podría reescribirse sin utilizar una función lambda de la siguiente manera:

````python
def get_second_element(t):
    return t[1] 
pairs.sort(key=get_second_element)
````

````python
pairs
````

Aquí usamos una lambda en una extracción de una lista (con la función `filter`).
La función `filter` toma dos argumentos: una función y un iterable. Aplica la función a cada elemento del iterable y devuelve un iterador con los elementos para los cuales la función devuelve `True`.

```{note}
Recuerda que un iterable es cualquier objeto en Python que puede ser recorrido elemento por elemento, como listas, tuplas, cadenas de texto, diccionarios y conjuntos.
```

````python
squares = [x**2 for x in range(100)]
sq = list(filter(lambda x : x%2 == 0 and x%3 == 0, squares))
sq
````
