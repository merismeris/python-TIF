# Funciones

Las funciones se usan para organizar el flujo del programa, especialmente para permitirnos realizar tareas comunes repetidamente de forma sencilla. Ya hemos usado muchas funciones, como las usadas a través del módulo `math`, integradas en métodos de clases como las que operan sobre listas (`append()` y `pop()`) o sobre cadenas (como `split()`). Aquí veremos cómo escribir nuestras propias funciones.

Para poder usar nuestras propias funciones, primero tenemos que definirlas. Podemos definirlas en el propio programa o en módulos externos que luego importamos. Veremos ambas opciones en prácticas. 

## Definición de funciones

Para definir una función debemos:
- Usar la palabra clave `def`, seguida del nombre de la función y paréntesis `()`.
- Dentro de los paréntesis, podemos listar los parámetros que la función tomará como entrada.
- Terminar la línea de definición con dos puntos `:`.
- Indentar el bloque de código que constituye el cuerpo de la función.
- Incluir una declaración de retorno para devolver un valor usando la palabra clave `return`. Si no se incluye una declaración de retorno, la función devolverá `None` por defecto.

Aquí se muestra un ejemplo de una función que toma un único argumento, `i`

````python
def my_fun(i):
    print(f"Recibido el valor i = {i}")
````

Para definir una función en un módulo externo, simplemente escribimos la definición de la función en un archivo `.py` separado. Por ejemplo, podemos crear un archivo llamado `mimodulo.py` con la siguiente función:

```python
def square(i):
    return i * i
```

### Llamada a funciones

Para llamar a la función, usamos su nombre seguido de paréntesis que contienen los argumentos que queremos pasarle. Por ejemplo:

````python
my_fun(10)
my_fun(5)
````

```{admonition} Nota
:class: note
Podemos definir una función en cualquier parte del código (siempre **antes de llamarla**), pero es una buena práctica definirlas al principio del archivo o en módulos separados para mantener el código organizado. 
```

Si hemos creado una función en un módulo externo, podemos importarla en nuestro programa principal usando la instrucción `import` o `from ... import ...`. 

En el ejemplo anterior, podemos importar la función `square` y usarla en nuestro programa principal de la siguiente manera:
```python
from mimodulo import square

print(square(5))  # Salida: 25
```

En cualquier caso, las funciones siempre devuelven un valor (o varios) de cualquier tipo. Si no se da explícitamente, devuelven `None` (el equivalente a `null` en otros lenguajes).

````python
a = my_fun(10)
a
````

A continuación se muestra una función que toma dos números y devuelve su producto.

````python
def multiply(a, b):
    return a * b

c = multiply(3, 4)
c
````

```{admonition} Nota
:class: note
Cada función trabaja dentro de su propio **ámbito** (_scope_). Cuando Python necesita un nombre dentro de una función, lo busca primero en ese ámbito; si no aparece, revisa los ámbitos que envuelven a la función (funciones anidadas), luego el módulo actual y, por último, los nombres incorporados del lenguaje. El lugar desde el que llamamos a la función no forma parte de esta cadena de búsqueda.

Para mantener el código claro, intenta que la función reciba todo lo que necesita a través de parámetros o cree sus propias variables. Depender de variables externas (variables globales) debe ser la excepción, no la norma.
```

````python
def saludar():
    print(f"Hola, {nombre}")  # Busca 'nombre'

# Contexto de la llamada #1
nombre = "María"
saludar()  # Imprime: Hola, María ✅

# Contexto de la llamada #2
nombre = "Pedro"
saludar()  # Imprime: Hola, Pedro ✅

# Pero esto NO funcionaría:
def otro_saludar():
    nombre = "Ana"
    saludar()  # ¿Usa la 'nombre' local de otro_saludar()?

otro_saludar()  # Imprime: Hola, Pedro (NO Ana)
                # Porque saludar() NO busca en otro_saludar()
````

## Paso de argumentos en Python

Tal como vimos en el tema 3, en Python, los objetos pueden ser mutables, si podemos modificar su valor después de ser creados, e inmutables, si no podemos hacerlo. Esta distinción es crucial porque afecta al comportamiento de las funciones cuando reciben argumentos.

### Modelo de paso de argumentos en Python

Python utiliza un modelo de **paso por asignación de objeto** (también llamado "paso por objeto" o "paso por compartición"). Esto significa que:

1. Cuando se llama a una función, Python pasa **referencias a los objetos**, no copias de los valores ni referencias a las variables.
2. El argumento de la función se convierte en un alias (otro nombre) que apunta al mismo objeto en memoria.

El comportamiento observable depende de si el objeto es mutable o inmutable:

- **Con objetos inmutables**: Si intentamos "modificar" el argumento dentro de la función, Python crea un nuevo objeto y el parámetro local apunta a este nuevo objeto. El objeto original fuera de la función permanece inalterado.
- **Con objetos mutables**: Si modificamos el contenido del objeto dentro de la función, estamos modificando el mismo objeto que existe fuera de la función, ya que ambos (el argumento y el parámetro) son referencias al mismo objeto en memoria.

A continuación se presenta un ejemplo para ilustrar este comportamiento:

````python
# Ejemplo con objeto mutable (lista)
def modificar_lista(lista):
    lista.append(4)  # Modifica el objeto original
    print(f"Dentro de la función: {lista}")

mi_lista = [1, 2, 3]
print(f"Antes de llamar: {mi_lista}")
modificar_lista(mi_lista)
print(f"Después de llamar: {mi_lista}")  # La lista original se modificó

print("\n" + "="*50 + "\n")

# Ejemplo con objeto inmutable (entero)
def modificar_entero(numero):
    numero = numero + 10  # Crea un nuevo objeto, no modifica el original
    print(f"Dentro de la función: {numero}")

mi_numero = 5
print(f"Antes de llamar: {mi_numero}")
modificar_entero(mi_numero)
print(f"Después de llamar: {mi_numero}")  # El entero original no cambió
````

```{admonition} Importante
:class: important

**Estrategias para evitar modificaciones no deseadas:**

- Hacer una copia explícita del objeto antes de modificarlo:
  - Para listas: `lista_copia = lista.copy()`
  - Para diccionarios: `dict_copia = dict_original.copy()`
  - Para copias profundas de objetos anidados: `import copy; copia = copy.deepcopy(objeto)`
- Utilizar tipos inmutables cuando sea posible (por ejemplo, tuplas en lugar de listas)
- Documentar claramente si una función modificará sus argumentos mutables

**Buena práctica:** Las funciones deberían evitar modificar sus argumentos mutables a menos que esa sea explícitamente su finalidad. En su lugar, es preferible devolver un nuevo objeto con las modificaciones.

Para más información, se puede consultar la [documentación oficial sobre el modelo de datos de Python](https://docs.python.org/3/reference/datamodel.html#objects-values-and-types).
```

## Argumentos por palabra clave

Es posible tener argumentos opcionales que proporcionen valores por defecto. A continuación se muestra una función que calcula $\sin(\theta)$ donde $\theta$ opcionalmente puede estar en grados.

````python
import math
````

````python
def my_sin(theta, in_degrees=False):
    if in_degrees:
        return math.sin(math.radians(theta))
    return math.sin(theta)
````

````python
my_sin(math.pi/4), my_sin(theta=45, in_degrees=True)
#my_sin(math.pi/4), my_sin(theta=45, in_degrees=True)     
#my_sin(math.pi/4), my_sin(in_degrees=True, theta=45)                 
````

```{admonition} Importante
:class: important
Los valores por defecto de los argumentos opcionales se evalúan **una única vez** cuando se define la función, no cada vez que se llama. Si el valor por defecto es un objeto mutable (como una lista o un diccionario), este mismo objeto se reutilizará en todas las llamadas a la función.

**Consecuencia:** Si se utiliza un objeto mutable vacío como valor por defecto (por ejemplo, `L=[]`), todas las llamadas a la función compartirán el mismo objeto, acumulando modificaciones entre llamadas. Este es uno de los errores más comunes entre programadores principiantes.

**Solución:** Utilizar `None` como valor por defecto y crear el objeto mutable dentro de la función cuando sea necesario.
```

A continuación se muestra un ejemplo que ilustra este problema común al usar una lista vacía como valor por defecto:

````python
def f(a, L=[]):
    L.append(a)
    return L
````

````python
print(f(1))
print(f(2))
print(f(3))
````

Obsérvese que cada llamada no crea su propia lista separada. En su lugar, **se creó una única lista vacía cuando la función fue procesada por primera vez, y esta lista persiste en memoria como el valor por defecto para el argumento opcional `L`**.

Si queremos que se cree una lista única cada vez (por ejemplo, un lugar separado en memoria), en su lugar **inicializamos el argumento a `None`** y luego comprobamos su valor real y creamos una lista vacía en el cuerpo de la función si el valor por defecto no cambió.

````python
def fnew(a, L=None):
    if L is None:
        L = []
    L.append(a)
    return L
````

````python
print(fnew(1))
print(fnew(2))
print(fnew(3))
````

También es posible pasar explícitamente una lista existente como argumento. En este caso, la función utilizará esa lista en lugar de crear una nueva, y las modificaciones se acumularán en la lista proporcionada:

````python
L = fnew(1)
print(fnew(2, L=L))
````

## Lambdas (funciones anónimas)

Las funciones lambda son funciones anónimas que se definen en una única línea. Son útiles cuando necesitamos una función simple y breve, especialmente para pasarla como argumento a otras funciones.

**Sintaxis básica:**
```python
lambda argumentos: expresión
```

Las lambdas pueden tener varios argumentos separados por comas, pero solo pueden contener una expresión que se devuelve automáticamente (no se usa `return`).

**Cuándo usar lambdas:**
- Cuando la función es muy simple y solo se necesita una vez
- Como argumento para funciones como `sort`, `filter`, `map`, etc.
- Para mantener el código más compacto y legible en casos sencillos

**Ejemplo:** Supongamos que tenemos una lista de tuplas y queremos ordenarla según el segundo elemento de cada tupla. El método `sort` acepta un argumento `key` que indica qué parte del elemento usar para ordenar:

````python
pairs = [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]
pairs.sort(key=lambda p: p[1])
````

````python
pairs
````

Las lambdas también son útiles con funciones como `filter` (que filtra elementos) o `map` (que aplica una operación a cada elemento). En este ejemplo, usamos `filter` con una lambda para seleccionar solo los cuadrados divisibles por 2 y por 3:

````python
squares = [x**2 for x in range(100)]
sq = list(filter(lambda x: x % 2 == 0 and x % 3 == 0, squares))
sq
````
