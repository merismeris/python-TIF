# Control de flujo

En algunas ocasiones, necesitamos que se ejecuten operaciones o tareas solo en determinados casos, bajo algunas condiciones. Por ejemplo, si un usuario tiene una edad mayor a 18 años, puede acceder a cierto contenido. Si no, no puede.

Para esto, Python nos ofrece las estructuras de **control de flujo** o sentencias de control al igual que sucede en otros lenguajes de programación. 

Las sentencias de control se pueden dividir principalmente en:

- **Sentencias condicionales (o selectivas):** permiten ejecutar un bloque de código solo si se cumple una condición. Son las sentencias `if`, `elif` y `else`.
- **Sentencias iterativas (también llamadas bucles):** permiten ejecutar un bloque de código de forma repetitiva. Cada repetición del cuerpo del bucle se denomina **iteración**. En Python, existen dos tipos de bucles: el bucle `while`, controlado por una condición lógica, y el bucle `for`, que recorre los elementos de un objeto iterable o un rango numérico.

```{admonition} Importante
:class: important
Python utiliza indentación para definir bloques de código. La recomendación de estilo es emplear cuatro espacios en blanco al inicio de cada línea indentada, pero lo fundamental es mantener una indentación consistente dentro del bloque.

```

## Bucle `while`

Permite ejecutar un bloque de código mientras una condición sea verdadera. La condición se evalúa **antes** de entrar en el bucle. Por tanto, se puede repetir entre 0 y n veces.

En el cuerpo se debe incluir alguna sentencia que en algún momento haga que la condición deje de cumplirse, para evitar bucles infinitos.

````python
n = 0
while n < 5:          # condición
    print(n)         # cuerpo del bucle
    n += 1       # actualización de la variable de control. Hemos usado el operador compacto `+=`: `n += 1` es lo mismo que `n = n + 1`.
````

```{admonition} Nota
:class: note
Observa que hay una variable (`n` en este caso) que se inicializa antes del bucle, se utiliza en la condición y se actualiza dentro del cuerpo del bucle. Esta variable se denomina variable de control del bucle.
```

## Bucle `For`

El bucle `for` permite iterar sobre una secuencia u objeto iterable, como una lista (`list`), una tupla (`tuple`), un diccionario (`dict`), un conjunto (`set`) o una cadena de texto (`str`) o sobre un rango de números. La variable de control del bucle toma cada uno de los valores de la secuencia o del rango en cada iteración, es decir, se convierte en un **iterador**. Internamente, Python obtiene un iterador mediante la función `iter()` y solicita valores sucesivos con `next()` hasta agotar la secuencia.

````python
alist = [1, 2.0, "three", 4]
for a in alist:
    print(a)
````

````python
for c in "this is a string":
    print(c)
````


```{admonition} Nota
:class: note
**Ojo** al usar range() para iterar de m hasta n, porque el valor n actúa como límite superior exclusivo. Si necesitas incluirlo, ajusta el rango a `range(m, n + 1)` y recuerda que el tercer argumento opcional define el tamaño del paso. `range(inicio, fin_exclusivo, paso)`, es decir, `for n in range(2, 10, 2):` iterará sobre los números pares del 2 al 8.
```

````python
for n in range(2, 10, 2):
    print(n)
````

````python
my_dict = {"key1":1, "key2":2, "key3":3}

for k, v in my_dict.items():
    print(f"key = {k}, value = {v}")
````

## Terminación de bucles 

Existen varias formas de terminar un bucle, entre las que se incluyen:

- Que la condición del bucle deje de cumplirse. Esto puede lograrse mediante:
    - Contador: inicializar una variable antes del bucle, utilizarla en la condición y actualizarla dentro del cuerpo (por ejemplo `contador += 1`).
    - Acumulador: emplear una variable para acumular un valor (suma, producto, etc.) y actualizarla en cada iteración.
    - Bandera lógica (flag): usar una variable booleana que se cambie dentro del bucle cuando se cumpla cierta condición.
- Que se produzca una excepción o error dentro del bucle (p. ej., mediante `try`/`except`). Nota: el manejo de errores se tratará aparte.
- Uso de la sentencia `break` dentro del cuerpo del bucle para salir de él de forma inmediata.

Consejo: prioriza condiciones claras y legibles; utiliza `break` con moderación cuando mejore la claridad del código.

### Sentencia `break`

La sentencia `break` permite salir de un bucle de forma inmediata, independientemente de si la condición del bucle es verdadera o falsa. Se utiliza dentro del cuerpo del bucle y puede ser útil para terminar un bucle cuando se cumple una condición específica. Si hay bucles anidados, `break` solo termina el bucle más interno en el que se encuentra.
```{admonition} Nota
:class: note
Usar `break` puede hacer que el código sea menos legible, por lo que se recomienda usarlo con moderación y solo cuando sea necesario.
```

````python
for i in range(10):
    if i == 5:
        break  # Sale del bucle cuando i es igual a 5
    print(i)
````

Ejemplo de bucles anidados donde `break` solo sale del bucle más interno

````python
for externo in range(3):
    print(f"Iniciando bucle externo {externo}")
    for interno in range(5):
        print(f"  iteración interna {interno}")
        if interno == 2:
            print("  -> encuentro condición, hago break (sale solo del bucle interno)")
            break  # rompe únicamente el bucle interno
    else:
        # Este else solo se ejecuta si el bucle interno terminó sin break
        print("  El bucle interno terminó normalmente (sin break)")
    print(f"Fin de la iteración externa {externo}\n")
````


### Sentencia `continue`

La sentencia `continue` permite saltar a la siguiente iteración del bucle, omitiendo el resto del código en el cuerpo del bucle para la iteración actual. Se utiliza dentro del cuerpo del bucle y puede ser útil para evitar ejecutar ciertas partes del código cuando se cumple una condición específica. Al igual que con `break`, si hay bucles anidados, `continue` solo afecta al bucle más interno en el que se encuentra.

```{admonition} Nota
:class: note
Al igual que con `break`, el uso excesivo de `continue` puede dificultar la lectura del código, por lo que se recomienda usarlo con moderación.
```

````python
for i in range(10):
    if i == 5:
        continue  # Omite el resto del cuerpo cuando i es igual a 5 y salta a la siguiente iteración.
    print(i)
````

### Sentencia `else` en bucles

La sentencia `else` puede utilizarse junto con los bucles `for` y `while`. El bloque de código dentro de `else` se ejecuta cuando el bucle termina normalmente, es decir, cuando la condición del bucle se vuelve falsa (en el caso de `while`) o cuando se han iterado todos los elementos (en el caso de `for`). Si el bucle se termina prematuramente mediante una sentencia `break`, el bloque `else` no se ejecuta.

```{admonition} Nota
:class: note
El uso de `else` con bucles puede mejorar la legibilidad del código al permitir separar claramente el código que se ejecuta después de que el bucle haya terminado de forma normal.
```

````python
for i in range(10):
    if i == 5:
        break  # Sale del bucle cuando i es igual a 5
    print(i)
else:
    print("Bucle terminado normalmente")
````

### Sentencia `pass`
La sentencia `pass` es una operación nula en Python. Se utiliza como un marcador de posición cuando se requiere una sintaxis pero no se desea ejecutar ninguna acción. En el contexto de los bucles, `pass` puede ser útil cuando se quiere definir un bucle que no realiza ninguna operación, pero que debe estar presente por razones de estructura del código o para futuras implementaciones.

```{admonition} Nota
:class: note
El uso de `pass` puede mejorar la legibilidad del código al indicar claramente que se ha considerado la necesidad de un bloque de código, pero que no se ha implementado ninguna acción en ese momento.
```
    

````python
n = 0
while n < 5:
    pass  # Bucle que no hace nada
    n += 1
````

## Sentencia `if-else`

La sentencia `if` permite ejecutar un bloque de código solo si se cumple una condición determinada. Si la condición es verdadera, se ejecuta el bloque de código indentado debajo de la sentencia `if`. Si la condición es falsa, se puede utilizar la sentencia `else` para ejecutar un bloque de código alternativo.
Adicionalmente, se puede utilizar la sentencia `elif` (abreviatura de "else if") para verificar múltiples condiciones de manera secuencial. Si la primera condición es falsa, se evalúa la siguiente condición `elif`, y así sucesivamente. Si ninguna de las condiciones es verdadera, se ejecuta el bloque de código bajo la sentencia `else`.
```python
if condición1:
    # Bloque de código si condición1 es verdadera
elif condición2:
    # Bloque de código si condición2 es verdadera
else:
    # Bloque de código si ninguna condición es verdadera
```


````python
x = 0

if x < 0:
    print("negativo")
elif x == 0:
    print("cero")
else:
    print("positivo")
````

Podemos anidar sentencias `if-else` dentro de otras sentencias `if-else` para manejar múltiples condiciones y casos complejos.

````python
var1=2
if var1>0:
    var1=var1+1
    if var1>=2:
        var1+=1
    else:
        var1=4
elif var1>-5:
    if var1==-3:
        var1-=1
    else:
        var1+=1
else:
    var1=0
````
